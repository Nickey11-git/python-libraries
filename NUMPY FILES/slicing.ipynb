# **Slicing**

```python
import numpy as np
```

```python
a1 = np.arange(10,30)

print(a1)
```

```python
a1[::-1] # reversing the array
```

```python
a1[::2] # alterante elements
```

```python
#output: [11,13,15]
a1[1:6:2]
```

```python
a1.size
```

# Factors of 20 ->
# (2,10), (4,5), (5,4), (10,20)

```python
a2 = np.reshape(a1,(4,5))
print(a2)
```

```python
a2[::2]
```

```python
a2[::2 , ::2]
```

```python
a2[::-2,::-3]
```

```python
a2[::2,::4]
```

```python
a2 = np.reshape(a1,(4,5))
print(a2)

a2[-2::-2,::-1]
#   a2[2::-2,::-1]  
```
