-------------------------------------------------------------------------------
arr1 shape : (2,3,2)

arr2 shape : (2,1,2) # here 1 is like VIP number

### **if there is a differnt shape, mismatching only 1 is allowed ->  supports broadcasting**
-------------------------------------------------------------------------------
arr1 shape : (2,3,2)

arr2 shape :   (3,2)
# or
arr1 shape : (4,2,3,2)

arr2 shape :     (3,2)

# 1.ARITHMATIC OPERATIONS

```python
# ARITHMATIC OPERATIONS
print(arr1 + arr2) # element by element operation is performed
print('-'*50)

print(arr1 - arr2) # element by element operation is performed
print('-'*50)

print(arr1 * arr2) # element by element operation is performed
print('-'*50)

print(arr1 / arr2) # element by element operation is performed
print('-'*50)

print(arr1 // arr2) # element by element operation is performed
print('-'*50)

print(arr1 % arr2) # element by element operation is performed
print('-'*50)

print(arr1 ** arr2) # element by element operation is performed
print('-'*50)
```

# 2.RELATIONAL OPERATOR


```python
# RELATIONAL OPERATOR
print(arr1 == arr2)
print('-'*50)
print(arr1 != arr2)
print('-'*50)
print(arr1 > arr2)
print('-'*50)
print(arr1 >= arr2)
print('-'*50)
print(arr1 < arr2)
print('-'*50)
print(arr1 <= arr2)
print('-'*50)
```

# 3.BITWISE OPERATORS

```python
# BITWISE OPERATORS
print(arr1 & arr2)
print('-'*50)

print(arr1 | arr2)
print('-'*50)

print(arr1 ^ arr2)
print('-'*50)

print(arr1 >> arr2)
print('-'*50)

print(arr1 << arr2)
print('-'*50)

print(~arr1)
print('-'*50)
```

# 3. **choice()** -> randomly it will pick the data.

# replace=False means it won't allow to generte duplicate values.

```python
c1 = rd.choice(['Red','Blue','Green','White','Black','Purple'])
print(c1)

print('-'*50)

c1 = rd.choice(['Red','Blue','Green','White','Black','Purple'],size=(3,))
print(c1)

print('-'*50)

c1 = rd.choice(['Red','Blue','Green','White','Black','Purple'],size=(3,) , p=[0.2,0.3,0.05,0.05,0.4,0])
print(c1)

print('-'*50)

c1 = rd.choice(['Red','Blue','Green','White','Black','Purple'],size=(3,) , p=[0.2,0.3,0.05,0.05,0.4,0] , replace=False)
print(c1)
```

# copy() and view() methods

## **copy()** ->  returns the shallow copy of the given array. means both the arrays are independent of each other

```python
a1 = np.array([10,20,30])
print(a1)
print('-'*50)


a2 = np.copy(a1)
print(a2)
print('-'*50)


a1[0] = 100
print(a1,a2,sep='\n')
```

## view() -> a1 and a2 are dependent on each other

```python
a1 = np.array([10,20,30])
print(a1)
print('-'*50)

a2 = a1.view()
print(a2)
print('-'*50)

a1[0] = 100
print(a1,a2,sep='\n')
```

## here reshape() -> here it returns copy

```python
a1 = np.arange(10,22)
print(a1,a1.size)
print('-'*50)

res = np.reshape(a1,shape=(2,6))
print(res)
print('-'*50)

a1[0] = 100
print(a1,res,sep='\n')

```

## ravel() -> it returns view

```python
a1 = np.arange(10,22)
print(a1,a1.size)
print('-'*50)

res = np.ravel(a1)
print(res)
print('-'*50)

a1[0] = 100
print(a1,res,sep='\n')

```

## here flatten() -> here it returns copy

```python
a1 = np.arange(10,22)
print(a1,a1.size)
print('-'*50)

res = a1.flatten()
print(res)
print('-'*50)

a1[0] = 100
print(a1,res,sep='\n')

```
