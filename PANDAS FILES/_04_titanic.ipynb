# **value_counts()** method which will will group and count one after the other

```python
df['ticket'].value_counts() #returns Series  (Occurance of each ticket)
# here index --> ticket data
# values ---> count of each ticket
```

```python
# Here we are converting series to dictionary
s = (df['ticket'].value_counts())
d = dict(s)
```

```python
df['passengers_count'] = df['ticket'].map(d)
```

```python
# Reamne the column
df = df.rename(columns = {'sex':'Gender','embarked':'Port'})
```

```python
df
```

```python
def companion_type_function(row):
    f=row['family_count']    #holds one value (single cell value)
    p=row['passengers_count']
    if p == 1:
        return 'solo'
    else:
        if f==0:
            return 'friends'
        else:
            return 'family'
            
   
    
```

## **apply()** is a method which acts like a intermediatary , it will go to dataframe and gives to function

```python
df['companion_type'] = df.apply(companion_type_function,axis=1)  # axis=1 means select all columns of 1st row,2nd row, .........., nth row)
```
