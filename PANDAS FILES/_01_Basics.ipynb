## Series() is a method which is used to create column for python collection


```python

# to make the data in a column wise
pd.Series(sal_list)
```

```python
#to reuse the data we need to store,so to store the data we need to assign a variable
sal = pd.Series(sal_list)
sal
```

```python
# to give a column name to the data
sal = pd.Series(sal_list, name='empsal')
sal
```

```python
# Here implicit conversion happen
sal_list=[1234,8000,4572,6546,8.7]
sal = pd.Series(sal_list, name='Empsal')  # name parameter is used to specify the column name
sal
```

```python
#accessing the cell value using key (INDEX)
sal[4]

# sal is a variable which holds thr column data (Series)
```

```python
print(sal[4])
```

```python
#updating/modifing the existing value
sal[4] = 9000
sal
```

```python
#sal[9]--->KeyError
sal[9] = 8989 #it creates new cell value bcz, index 9 doesn't exist
sal
```

```python
#to remove the data which we are specified
sal.drop(5)
```

```python
sal
```

## **drop()** is a method which is used to remove the cell based on given index/key.

```python
# to make the data permanent after removing the data, do reinitialization 
sal = sal.drop(5)
sal
```

```python
sal
```

### **reset_index()** is a method which is used to reset the old_index to default_index[0,1,2...,n]


```python
sal.reset_index()
```

## **(drop = True)** can be used to avoid the old_index as new_column


```python
sal.reset_index(drop=True)
```

### **astype** is a method which is used to perform explicit type conversion for column

```python
# explicit type conversion in pandas
sal.astype('int')
```

```python
sal
```

```python
# we did converion permanent by assigning variable to it.
sal = sal.astype('int')
sal
```
