## 28-08-2026 
->Pandas stands for **Paneldata** (tabular format/structured data)

->it has two data structures
1. Series
2. DataFrame

**Series**: 
->it represents column. 

->column should be homogeneous type of data.

->column can have n number of values.

->cell value can be accessing using **index**.

->it supports duplicates.

->it is an **ordered collection**.

->it is mutable(means we can perform DML operations).


**DataFrame**: ->it represents table. 

->it supports homogeneous or heterogeneous data.

->cell value can be accessing using **row index and column labels**.

->it supports **duplicates rows not duplicate columns**.

->it is an **ordered collection**.

->it is mutable(means we can perform DML operations).





list is used for creating series
dictionary is used in creating tables in DataFrame
